```python
# access modifiers
# getters and setters methods

# class User:
#     def __init__(self, username, email, password):
#         self.username= username
#         self.email = email
#         self.password = password
# lets try to make a data attribute protected (email for example ) 
# self.email = email becomes self._email = email

class User:
    def __init__(self, username, email, password):
        self.username= username
        self._email = email    # cant be accessed out of this class or in any subclass
        self.password = password
    def get_email(self):# this gives us more control over the data and how it should be displayed 
        return self._email
    def set_email(self, newmail): # cant be set to anything so wwe have control over it 
        if '@' in newmail:
            self._email = newmail
        
        
ana = User('ely', 'ely.com', 'abc')
# so for protected data we use getters and setters to access them and modify them
print(ana.get_email())
ana.set_email('nfuewrfuw@dd.com')
ana.get_email()
```

```python
# better way using properties 
class User:
    def __init__(self, username, email, password):
        self.username= username
        self._email = email
        self.password = password
    @property
    def email(self):
        print(' email accessed ')
        return self._email
    # the property thing basically makes calling the function email like an object without the parenthesis
    
    # now the setter, we use the previous property for the settter
    @email.setter 
    def email(self, value):
        self._email = value 
    
ana = User('ely', 'ely.com', 'abc')
print(ana.email)
ana.email= 'ana@gmail.com'
print(ana.email)
```

```python
# static attributes
# its an attribute that belongs to the class rather than individual objects
class User:
    user_count = 0
    def __init__(self, username, email, password):
        self.username= username
        self._email = email
        self.password = password
        User.user_count += 1 

    def __repr__(self):
        return f'username is : {self.username}, password : {self.password}'
ana = User('ely', 'ely.com', 'abc')
howa = User('adn','adn.com','obito')
print(howa)
print(User.user_count)
```

```python
#static methods 

class BankAcc:
    
    MIN_BALANCE = 100

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance 

    def deposit(self, amount):
        if self._is_valid_amount(amount):
            self.balance += amount
            print(self.__transaction_log('deposit', amount))
        else:
            print(f'the deposited amount must be positive')

    def _is_valid_amount(self,amount):
        return amount> 0
    def __transaction_log(self, transaction_type, amount):
        return f"Logging {transaction_type} of ${amount}"
    @staticmethod    # we can use this even if we dont have any object.... #memory efficiency# 
    def is_valid_interest_rate(rate):
        return 0<rate<5

account = BankAcc('alice', 500)
account.deposit(200)
BankAcc.is_valid_interest_rate(3)
```

```python

# OOP PRINCIPLES 

# ENCAPULATION basically control the access to the class, [program can't be abused by users]

class BankAcc : 
    def __init__(self):
        self._balance = 0.0

    @property
    def balance(self):
        return self._balance

    def deposit(self,amount):
        if 0< amount:
            self._balance += amount
    def withdraw(self,amount):
        if 0<amount<self._balance:
            self._balance -= amount 
        else:
            raise ValueError("amount is not valid")

luka = BankAcc()
luka.balance
luka.deposit(200)
luka.withdraw(250)
luka.balance



# so encapsulation provides controlled access to the data 

```
